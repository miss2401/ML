```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.spatial.distance import cdist
import random

np.random.seed(18)
```

```python
means= [[2,2], [8,3], [3, 6]] # Tao 3 ky vong ung voi 3 clusters trong ma tran ki vong
cov = [[1, 0], [0, 1]] #Ma tran hiep phuong sai
N = 500 # So luong diem du lieu
X0 = np.random.multivariate_normal(means[0], cov, N)
X1 = np.random.multivariate_normal(means[1], cov, N)
X2 = np.random.multivariate_normal(means[2], cov, N)# Tao ma tran voi 500 diem du lieu tuan theo phan phoi chuan da chieu
```

```python
X = np.concatenate((X0, X1, X2), axis=0) # ghep 3 ma tran tren theo chieu doc tao thanh ma tran 1500 diem du lieu
K = 3
original_label = np.asarray([0]*N + [1]*N + [2]*N).T #vecto do dai 1500,
print(original_label.shape)
```

```python
def Kmeans_init_centroids(X, k):
    return X[np.random.choice(X.shape[0], k, replace = False)] #Chọn ngẫu nhiên k điểm ra làm centroids
def Kmeans_assign_labels(X, centroids):
    D = cdist(X, centroids) # Tinh khoang cach tung diem trong X so voi diem centroid => tra ve ma tran kich thuoc 1500, 3, 3 cot ung voi 3 khoang cach den 3 centroid
    return np.argmin(D, axis = 1) #Tra ve vi tri cua diem gan voi centroid nhat 0, 1, 2 tren tung hang
def has_converged(centroids, new_centroids):
    return (set([tuple(a) for a in centroids]) == set([tuple(a) for a in new_centroids])) # so sanh tung toa do mot cua moi centroid trong list centroids
def Kmeans_update_centroids(X, labels, K):
    centroids = np.zeros((K,X.shape[1])) # tao ma tran chua cac centroi kich thuoc 3x2
    for k in range(K):
        Xk = X[labels == k, :] #Tap hop nhung diem ma thuoc cluster thu k o tren da xet
        centroids[k, :] = np.mean(Xk, axis = 0) # Tinh centroid moi bang cach lay trung binh cac diem da chon voi Y co dinh => tim m
    return centroids
```

```python
def Kmeans(X, k):
    centroids = [Kmeans_init_centroids(X, k)] # sorry nham, o day se la mot list chua bo 3 centroids
    labels = []
    it = 0
    #Hiểu nôm na cái vòng lặp này là no có một bộ centroids ứng vs labels và sau mỗi vòng lặp nó sẽ tính lại centroids từ đó suy ra labels mới. labels ở vòng lặp đầu tiên có sẵn rồi
    while(1):
        labels.append(Kmeans_assign_labels(X, centroids[-1])) # them bo cac vi tri cua tung diem khi so khoang cach voi bo centroids moi nhat
        new_centroids = Kmeans_update_centroids(X, labels[-1], k)
        if(has_converged(centroids[-1], new_centroids)):
            break
        centroids.append(new_centroids)
        it+=1
    return (centroids, labels, it)
```

```python
(centroid, labels, it) = Kmeans(X, K)
print(centroid[-1]) # in ra 3 centroids xấp xỉ đúng
```
