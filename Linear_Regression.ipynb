```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model, datasets
from numpy.ma.core import concatenate
```

```python
X = np.array([[147, 150, 153, 158, 163, 165, 168, 170, 173, 175, 178, 180, 183]]).T # Ma trận chuyển vị từ hangf sang cột 13 x 1
print(X.shape)
y = np.array([ 49, 50, 51, 54, 58, 59, 60, 62, 63, 64, 66, 67, 68]) #vectỏr kich thuoc 13,
```

```python
one = np.ones((X.shape[0], 1))#Tao ma tran don vi kich thuoc 13 x 1 bang X
Xbar = np.concatenate((one, X), axis=1).T  #Noi ma tran don vi o tren voi ma tran X => gia tri dau tien x0 luon bang 1 o moi vecto => bias trick => kich thuoc 13x2 vaf chuyen vi để coi mỗi điểm dữ liệu là 1 cột2 x 13
A = np.dot(Xbar, Xbar.T) #=> kich thuoc 2x2
b = np.dot(Xbar, y) #=> tinh Xy kich thuoc  2,
w = np.dot(np.linalg.inv(A), b) # tinh vecto trong so w kich thuoc 2,
w_0, w_1 = w[0], w[1]
print(w)
```

```python
y1 = w_1 * 155 + w_0
y2 = w_1 * 160 + w_0
print("Input 155cm, true output 52kg, predicted output %.2fkg "%(y1) )
print("Input 160cm, true output 56kg, predicted output %.2fkg" %(y2) )
```

```python
plt.plot(X, y, 'ro')
y_pred = w_0 + X * w_1
plt.plot(X, y_pred, 'b')
plt.axis([140, 190, 45, 75])
plt.xlabel('Height')
plt.ylabel('weight')
plt.show()
```

```python
regr = linear_model.LinearRegression()
regr.fit(X, y)
w_1 = regr.coef_[0]
w_0 = regr.intercept_
y_pred = w_0 + X * w_1
y_pred_first = w[0] + X * w[1]
plt.plot(X, y_pred_first, 'g--')
plt.plot(X, y, 'ro')
plt.plot(X, y_pred, 'b')
plt.xlabel("Height")
plt.ylabel("weight")
plt.axis([140, 190, 45, 75])
plt.show()
```
