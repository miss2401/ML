```python
import numpy as np
from time import time
```

```python
d, N = 1000, 10000
X = np.random.randn(N, d) # Tạo N điểm có d chieu hay ma trận gồm N hàng và d cột
z = np.random.randn(d) # Tạo vecto kich thuoc d
```

```python
def dist_pp(z, x):
    d = z - x.T # tinh theo cong thuc do dai euclid
    return np.sum(d*d)
def dist_ps_naive(z, X):
    res = np.zeros((1, N)) #tao ma tran kich thuoc 1 x N de luu cac gia tri do dai euclid giua cac diem trong ma tran voi diem z
    for i in range(N):
        res[0][i] = dist_pp(z, X[i])
    return res
def dist_ps_fast(z, X):
    x2 = np.sum(X**2, axis=1) # Tinh tong binh phuong theo tung hang
    z2 = np.sum(z**2) # tinh binh phuong do dai euclid
    return x2 + x2 - 2*X.dot(z)
```

```python
def dist_pp(z, x):
    d = z - x.reshape(z.shape) # force x and z to have the same dims
    return np.sum(d*d)
# from one point to each point in a set, naive
def dist_ps_naive(z, X):
    N = X.shape[0]
    res = np.zeros((1, N))
    for i in range(N):
        res[0][i] = dist_pp(z, X[i])
    return res
# from one point to each point in a set, fast
def dist_ps_fast(z, X):
    X2 = np.sum(X*X, 1) # square of l2 norm of each ROW of X
    z2 = np.sum(z*z) # square of l2 norm of z
    return X2 + z2 - 2*X.dot(z) # z2 can be ignored
```

```python
# naively compute square distance between two vector

t1 = time()
D1 = dist_ps_naive(z, X)
print("naive point2set, running time:", time() - t1, "s")
t1 = time()
D2 = dist_ps_fast(z, X)
print("fast point2set , running time:", time() - t1, "s")
print("Result difference:", np.linalg.norm(D1 - D2))
```

```python
# Tinh khoang cach giua cac cap diem du lieu trong 2 ma tran
M = 100
Z = np.random.randn(M, d)
def dist_ss_0(Z, X):
    res = np.zeros((M, N)) #Tao ma tran 0 kich thuoc M x N, Z kich thuoc M x d, X kich thuoc N x d => khac ve so luong diem
    for i in range(M):
        res[i] = dist_ps_naive(Z[i], X)
    return res
def dist_ss_fast(Z, X):
    X2 = np.sum(X**2, axis = 1)
    Z2 = np.sum(Z**2, axis = 1)
    return X2.reshape(-1, 1) + Z2 .reshape(1, -1) - 2*X.dot(Z)
t1 = time()
D3 = dist_ss_0(Z, X)
print("half fast set2set running time:", time() - t1)
t1 = time()
D4 = dist_ss_fast(Z, X)
print("fast set2set running time", time() - t1)
print("Result difference:", np.linalg.norm(D3 - D4))
```
