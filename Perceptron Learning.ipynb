```python
import numpy as np
```

```python
def predict(w, X):
    return np.sign(X.dot(w))
def perceptron(X, y, w_init):
    w = w_init
    while 1:
        pred =  predict(w, X)
        mis_idx = np.where(np.equal(y, pred) == 0)[0]
        num_mis = mis_idx.shape[0]
        if num_mis == 0:
            return w
        random_id = np.random.choice(mis_idx, 1)[0]
        w = w + y[random_id] * X[random_id]
```

```python
means = [[-1, 0], [1, 0]] # Khởi tạo ma trận kì vong
cov = [[.3, .2], [.2, .3]] # Ma trận hiệp phương sai
N = 10 # Số lượng điểm dữ liệu
X0 = np.random.multivariate_normal(means[0], cov, N) # Khởi tạo ma trận chứa mỗi hàng là 1 điểm dữ liệu tuân theo phân phối chuẩn
X1 = np.random.multivariate_normal(means[1], cov, N)
print(X0.shape)
X = np.concatenate((X0, X1), axis = 0) # Nối X0 và X1 theo chiều dọc => kích thước (2N, 2)

y = np.concatenate((np.ones(N), -1*np.ones(N))) # Gán nhãn 1 và -1 tương ứng
Xbar = np.concatenate((np.ones((2*N, 1)), X), axis = 1) # khởi tạo ma trận với cột đầu là giá trị 1 và bên cạnh là ma trận vì có bias
w_init = np.random.randn(Xbar.shape[1])# Tạo vector w với kích thước 3,
w = perceptron(Xbar, y, w_init)
```
